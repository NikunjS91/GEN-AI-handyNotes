```python
from pydantic import BaseModel,Field
from typing import List,Annotated
```

```python
class user(BaseModel):
    name : str
    age : int
```

```python
User = user(name= "Nikunj",age=27)
User
```

### Field validation

```python
class product(BaseModel):
    title : str = Field(max_length=10)
    desc : str = Field(min_length=20)
    price : int = Field(gt=0,le=100) ## gt stands for grater then and le less then
```

```python
Product = product(title="samsung",desc="samsung is a good comapany and they make good phones",price=88)
Product2 = product(title="Apple",desc="Apple is a good comapany and they make good phones",price=99)
Product
```

### multiple class validation

```python
class orde(BaseModel):
    id :str = Field(max_length=12,min_length=6)
    product : product
```

```python
Order = orde(
    id = "1234567H",
    product =Product ## either this of pass a dictionary with key-pair matching the rules for validation.
)
```

```python
Order
```

```python
Order.product.title
```

```python
class Carts(BaseModel):
    id : str =Field(min_length=6,max_length=12)
    items: List[product]
    comment :str = Field(
        default="No Comment"
        )
```

```python
cart=Carts(id="123456qw",items=[Product,Product2],comment="nice product")
cart
```

```python
for p in cart.items:
    print(p.title)
```

### annotated allows use to create common validator or create custom validator.

```python
commonValidator = Annotated[List[product],Field(default=[])]
```

```python
class Carts(BaseModel):
    id : str =Field(min_length=6,max_length=12)
    items: List[product]
    comment :str = Field(
        default="No Comment"
        )
```

```python
class Order(BaseModel):
    product : commonValidator
```

```python
ord = Order(product=[Product,Product2])
```

```python
ord.product
```

```python
for pro in ord.product:
    print(pro.title)
```
